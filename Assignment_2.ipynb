```python

def rol(binary, positions):
    print("Original number:", binary)

    for i in range(positions):
        first_bit = binary[0]
        binary = binary[1:] + first_bit
        print("Step", i + 1, ":", binary)

    print("Final ROL result:", binary)

def rcl(binary, positions, carry):
    print("Original number:", binary)
    print("Initial carry:", carry)

    for i in range(positions):
        first_bit = binary[0]
        binary = binary[1:] + str(carry)
        carry = int(first_bit)

        print("Step", i + 1, ":", binary,
              "Carry:", carry)

    print("Final RCL result:", binary)
    print("Final carry:", carry)

print("a) Rotate Left without Carry (ROL)")
rol("11110110", 2)

print("\nb) Rotate Left through Carry (RCL)")
rcl("10000110", 2, 0)
```
