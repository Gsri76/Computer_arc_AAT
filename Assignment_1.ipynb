```python

def to_binary(n, bits):
    return format(n & ((1 << bits) - 1), f"0{bits}b")

def signed_value(binary):
    bits = len(binary)
    value = int(binary, 2)

    if binary[0] == '1':
        value -= (1 << bits)

    return value

def calculate(a, b):
    print("\nFirst operand :", to_binary(a, 8))
    print("Second operand:", to_binary(b, 8))

    # Perform subtraction using 16-bit 2's complement
    a16 = to_binary(a, 16)
    negative_b16 = to_binary(-b, 16)

    print("16-bit A      :", a16)
    print("16-bit -B     :", negative_b16)

    result16 = a - b
    binary16 = to_binary(result16, 16)

    print("16-bit Result :", binary16)
    print("Decimal Result:", result16)

    # Check signed 8-bit overflow
    overflow = result16 < -128 or result16 > 127

    if overflow:
        print("8-bit Overflow: YES")
    else:
        print("8-bit Overflow: NO")

    print("8-bit Result  :", to_binary(result16, 8))

print("a) 35 - (-29)")
calculate(35, -29)

print("\nb) (-35) - (-12)")
calculate(-35, -12)
```
